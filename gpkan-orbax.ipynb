```python
from gpkanmodel.model import GPKAN
import orbax.checkpoint as ocp
import jax
model = GPKAN()

model_params = {
    "latent_grid": model.latent_grids,
    "latent_support": model.latent_supports,
    "kernel_params": model.kernel_parameters,
    }
```

```python
from pathlib import Path
path_str = Path('checkpoints/gpkan_params').absolute()
path = ocp.test_utils.erase_and_create_empty(path_str)
```

```python
ckptr = ocp.StandardCheckpointer()
# 3. Save the parameters
# 'args' expects an ocp.args.StandardSave(item)
ckptr.save(
    path/"poop",
    model_params
)
```

```python
restored_params = ckptr.restore(
    path/"poop",
    model_params
)
```
